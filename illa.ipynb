```python
import matplotlib.pyplot as plt

years = [2022, 2023, 2024, 2025]

python = [20, 30, 40, 50]
sql = [30, 35, 40, 45]
powerbi = [10, 20, 30, 40]

# Stack plot
plt.stackplot(
    years,
    python,
    sql,
    powerbi,
    labels=["Python", "SQL", "Power BI"]
)

# Data labels
for i, year in enumerate(years):

    # Python
    plt.text(
        year,
        python[i] / 2,
        str(python[i]),
        ha="center"
    )

    # SQL
    plt.text(
        year,
        python[i] + sql[i] / 2,
        str(sql[i]),
        ha="center"
    )

    # Power BI
    plt.text(
        year,
        python[i] + sql[i] + powerbi[i] / 2,
        str(powerbi[i]),
        ha="center"
    )

# Title and axis labels
plt.title("Learning Progress")
plt.xlabel("Year")
plt.ylabel("Learning Level")

# Legend
plt.legend()

plt.show()
```

```python
import tkinter as tk
from tkinter import messagebox

def analyze():
    try:
        name = name_entry.get()
        roll_no = roll_entry.get()

        m1 = float(mark1_entry.get())
        m2 = float(mark2_entry.get())
        m3 = float(mark3_entry.get())
        m4 = float(mark4_entry.get())
        m5 = float(mark5_entry.get())

        total = m1 + m2 + m3 + m4 + m5
        percentage = total / 5

        if percentage >= 90:
            grade = "A+"
        elif percentage >= 80:
            grade = "A"
        elif percentage >= 70:
            grade = "B"
        elif percentage >= 60:
            grade = "C"
        elif percentage >= 50:
            grade = "D"
        else:
            grade = "F"

        if percentage >= 50:
            result = "PASS"
        else:
            result = "FAIL"

        result_label.config(
            text=f"Name: {name}\n"
                 f"Roll No: {roll_no}\n"
                 f"Total: {total}/500\n"
                 f"Percentage: {percentage:.2f}%\n"
                 f"Grade: {grade}\n"
                 f"Result: {result}"
        )

    except ValueError:
        messagebox.showerror("Error", "Please enter valid marks.")


# Create window
window = tk.Tk()
window.title("Student Performance Analyzer")
window.geometry("400x550")

# Heading
title = tk.Label(
    window,
    text="Student Performance Analyzer",
    font=("Arial", 18, "bold")
)
title.pack(pady=15)

# Student name
tk.Label(window, text="Student Name").pack()
name_entry = tk.Entry(window)
name_entry.pack()

# Roll number
tk.Label(window, text="Roll Number").pack()
roll_entry = tk.Entry(window)
roll_entry.pack()

# Marks
tk.Label(window, text="Subject 1 Marks").pack()
mark1_entry = tk.Entry(window)
mark1_entry.pack()

tk.Label(window, text="Subject 2 Marks").pack()
mark2_entry = tk.Entry(window)
mark2_entry.pack()

tk.Label(window, text="Subject 3 Marks").pack()
mark3_entry = tk.Entry(window)
mark3_entry.pack()

tk.Label(window, text="Subject 4 Marks").pack()
mark4_entry = tk.Entry(window)
mark4_entry.pack()

tk.Label(window, text="Subject 5 Marks").pack()
mark5_entry = tk.Entry(window)
mark5_entry.pack()

# Analyze button
tk.Button(
    window,
    text="Analyze Performance",
    command=analyze
).pack(pady=15)

# Result
result_label = tk.Label(
    window,
    text="",
    font=("Arial", 12),
    justify="left"
)
result_label.pack(pady=10)

window.mainloop()
```
