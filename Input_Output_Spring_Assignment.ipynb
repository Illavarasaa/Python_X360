```python
#30. Input → Process → Output Challenge

#QUESTION
#Create your own small Python program that takes at least three inputs, 
#processes the strings using at least three different string methods, 
#and produces a properly formatted output using an f-string.

#ANSWER

name = input("Enter your name: ")
city = input("Enter your city: ")
hobby = input("Enter your hobby: ")

name = name.strip().title()
city = city.strip().upper()
hobby = hobby.strip().lower()

print(f"Hello {name}! You live in {city} and enjoy {hobby}.")
```

```python
#29. String Transformation

#QUESTION
#Ask the user to enter a sentence and produce a new version where the sentence is cleaned, 
#converted to lowercase, spaces are replaced with hyphens, and the final result is displayed in reverse.

#ANSWER

sentence = input("Enter a sentence: ")

result = sentence.strip().lower().replace(" ", "-")
print("Transformed and reversed:", result[::-1])
```

```python
#27. Email Analyzer

#QUESTION
#Ask the user to enter an email address. 
#Display whether the email starts with a valid-looking username, 
#whether it ends with a common domain such as .com, and the position of the @ symbol.

#ANSWER

email = input("Enter an email address: ").strip()

username = email.split("@")[0] if "@" in email else ""
valid_username = bool(username) and " " not in username
common_domain = email.lower().endswith(".com")
at_position = email.find("@")

print("Valid-looking username:", valid_username)
print("Ends with .com:", common_domain)
print("Position of @:", at_position)
```

```python
#26. Mini Text Processor

#QUESTION
#Ask the user to enter a sentence. Clean the sentence, convert it to lowercase, 
#count the number of words, replace spaces with hyphens, and display the reversed sentence.

#ANSWER

sentence = input("Enter a sentence: ")

cleaned = sentence.strip().lower()
word_count = len(cleaned.split())
hyphenated = cleaned.replace(" ", "-")
reversed_sentence = cleaned[::-1]

print("Cleaned:", cleaned)
print("Word count:", word_count)
print("Hyphenated:", hyphenated)
print("Reversed:", reversed_sentence)
```

```python
#5. Shopping Bill

#QUESTION
#Ask the user to enter the price of one item and the quantity purchased. 
#Calculate and display the total amount with two decimal places.

#ANSWER

price = float(input("Enter the price of one item: "))
quantity = int(input("Enter the quantity purchased: "))

total = price * quantity
print(f"Total amount: ${total:.2f}")
```

```python
#3. Simple Calculator

#QUESTION
#Ask the user to enter two numbers. Display their sum, difference, product, and quotient.

#ANSWER

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)

if b != 0:
    print("Quotient:", a / b)
else:
    print("Quotient: Cannot divide by zero.")
```
