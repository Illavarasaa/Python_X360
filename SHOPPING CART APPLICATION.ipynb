```python
# SHOPPING CART APPLICATION

# QUESTION
#Create a shopping cart GUI where users can add products and prices, manage the cart and generate a bill.
#The GUI should contain Product Name, Product Price, Add Product, Remove Selected, Sort Products,
#Generate Bill, and Total/Discount/Final Amount.

#ANSWER


import tkinter as tk
from tkinter import messagebox


# Create the main window
root = tk.Tk()
root.title("Shopping Cart")
root.geometry("600x500")


# Lists to store product names and prices
products = []
prices = []


# ---------------------------------------------------------
# REFRESH CART
# ---------------------------------------------------------

def refresh_cart():

    # Clear the existing items from the Listbox
    cart_list.delete(0, tk.END)

    # Display all products and their prices
    for i in range(len(products)):
        cart_list.insert(
            tk.END,
            f"{products[i]} - ${prices[i]:.2f}"
        )


# ---------------------------------------------------------
# ADD PRODUCT
# ---------------------------------------------------------

def add_product():

    # Get product name from the Entry box
    name = product_entry.get().strip()

    try:
        # Get product price and convert it to float
        price = float(price_entry.get())

        # Price cannot be negative
        if price < 0:
            raise ValueError

    except ValueError:
        # Display error if price is invalid
        messagebox.showerror(
            "Error",
            "Enter a valid product price."
        )
        return

    # Check whether product name is empty
    if name == "":
        messagebox.showerror(
            "Error",
            "Enter a product name."
        )
        return

    # Add product and price to the lists
    products.append(name)
    prices.append(price)

    # Update the cart
    refresh_cart()

    # Clear input boxes
    product_entry.delete(0, tk.END)
    price_entry.delete(0, tk.END)


# ---------------------------------------------------------
# REMOVE SELECTED PRODUCT
# ---------------------------------------------------------

def remove_selected():

    # Get the selected item from the Listbox
    selected = cart_list.curselection()

    # Check whether an item is selected
    if not selected:
        messagebox.showwarning(
            "Warning",
            "Select a product to remove."
        )
        return

    # Get the index of selected product
    index = selected[0]

    # Remove product and corresponding price
    products.pop(index)
    prices.pop(index)

    # Refresh the cart
    refresh_cart()


# ---------------------------------------------------------
# SORT PRODUCTS
# ---------------------------------------------------------

def sort_products():

    # Combine product names and prices
    combined = list(zip(products, prices))

    # Sort products alphabetically
    combined.sort(
        key=lambda item: item[0].lower()
    )

    # Clear the original lists
    products.clear()
    prices.clear()

    # Add sorted products back to the lists
    for name, price in combined:
        products.append(name)
        prices.append(price)

    # Refresh the cart
    refresh_cart()


# ---------------------------------------------------------
# GENERATE BILL
# ---------------------------------------------------------

def generate_bill():

    # Calculate total price
    total = sum(prices)

    # Calculate discount
    if total >= 1000:
        discount = total * 0.10

    elif total >= 500:
        discount = total * 0.05

    else:
        discount = 0

    # Calculate final amount
    final_amount = total - discount

    # Display the bill
    bill_text.set(
        f"Total: ${total:.2f}\n"
        f"Discount: ${discount:.2f}\n"
        f"Final Amount: ${final_amount:.2f}"
    )


# ---------------------------------------------------------
# CLEAR / RESET
# ---------------------------------------------------------

def clear_all():

    # Clear products and prices
    products.clear()
    prices.clear()

    # Clear Listbox
    cart_list.delete(0, tk.END)

    # Clear bill
    bill_text.set("")

    # Clear input boxes
    product_entry.delete(0, tk.END)
    price_entry.delete(0, tk.END)


# =========================================================
# GUI DESIGN
# =========================================================

# Application heading
tk.Label(
    root,
    text="Shopping Cart",
    font=("Arial", 16, "bold")
).pack(pady=10)


# Product Name
tk.Label(
    root,
    text="Product Name"
).pack()

product_entry = tk.Entry(root)
product_entry.pack()


# Product Price
tk.Label(
    root,
    text="Product Price"
).pack()

price_entry = tk.Entry(root)
price_entry.pack()


# Add Product button
tk.Button(
    root,
    text="Add Product",
    command=add_product
).pack(pady=5)


# Shopping cart Listbox
cart_list = tk.Listbox(
    root,
    width=50,
    height=10
)
cart_list.pack(pady=10)


# Remove Selected button
tk.Button(
    root,
    text="Remove Selected",
    command=remove_selected
).pack(pady=3)


# Sort Products button
tk.Button(
    root,
    text="Sort Products",
    command=sort_products
).pack(pady=3)


# Generate Bill button
tk.Button(
    root,
    text="Generate Bill",
    command=generate_bill
).pack(pady=3)


# Variable used to display the bill
bill_text = tk.StringVar()


# Display Total, Discount and Final Amount
tk.Label(
    root,
    textvariable=bill_text,
    justify="left",
    font=("Arial", 11)
).pack(pady=10)


# Clear / Reset button
tk.Button(
    root,
    text="Clear / Reset",
    command=clear_all
).pack(pady=5)


# Start the Tkinter event loop
root.mainloop()
```
