```python
#1. STUDENT PERFORMANCE ANALYZER

#QUESTION
#Create a GUI application that accepts student details and generates a complete performance report. 
#The GUI should contain Student Name, Number of Subjects, an "Enter Marks" button, dynamically generated 
#mark input fields, a "Calculate Result" button, and a result area showing Total, Average, Pass/Fail and Grade.

#ANSWER

import tkinter as tk
from tkinter import messagebox

# Create the main Tkinter window
root = tk.Tk()

# Set the title of the application window
root.title("Student Performance Analyzer")

# Set the size of the application window
root.geometry("500x600")


# List used to store all mark Entry widgets
mark_entries = []


# ---------------------------------------------------------
# FUNCTION 1: ENTER MARKS
# ---------------------------------------------------------

def enter_marks():
    # Tell Python that we want to modify the global mark_entries list
    global mark_entries

    # Remove previously created subject/mark widgets
    for widget in marks_frame.winfo_children():
        widget.destroy()

    # Reset the mark entries list
    mark_entries = []

    # Try to get and validate the number of subjects
    try:
        # Get the value entered by the user and convert it to integer
        n = int(subject_entry.get())

        # Number of subjects must be greater than zero
        if n <= 0:
            # Manually raise ValueError if the number is invalid
            raise ValueError

    # Handle invalid input
    except ValueError:
        # Display an error message to the user
        messagebox.showerror(
            "Error",
            "Enter a valid positive number of subjects."
        )

        # Stop this function
        return

    # Create an Entry box for each subject
    for i in range(n):

        # Create a label such as "Subject 1 Mark:"
        tk.Label(
            marks_frame,
            text=f"Subject {i + 1} Mark:"
        ).pack()

        # Create a text box where the user can enter the mark
        entry = tk.Entry(marks_frame)

        # Display the Entry box
        entry.pack()

        # Store the Entry widget in the list
        mark_entries.append(entry)


# ---------------------------------------------------------
# FUNCTION 2: CALCULATE RESULT
# ---------------------------------------------------------

def calculate_result():

    # Check whether the user has created mark input boxes
    if not mark_entries:

        # Show an error message if marks have not been entered
        messagebox.showerror(
            "Error",
            "Click 'Enter Marks' first."
        )

        # Stop the function
        return

    # Try to read and validate all marks
    try:

        # Create an empty list to store marks
        marks = []

        # Go through every mark Entry widget
        for entry in mark_entries:

            # Get the entered value and convert it to float
            mark = float(entry.get())

            # Marks must be between 0 and 100
            if mark < 0 or mark > 100:

                # Raise ValueError if the mark is invalid
                raise ValueError

            # Add the valid mark to the marks list
            marks.append(mark)

    # Handle invalid marks
    except ValueError:

        # Show an error message
        messagebox.showerror(
            "Error",
            "Enter marks between 0 and 100."
        )

        # Stop the function
        return


    # Calculate the total marks
    total = sum(marks)

    # Calculate the average marks
    average = total / len(marks)


    # -----------------------------------------------------
    # CHECK PASS OR FAIL
    # -----------------------------------------------------

    # Check whether any subject mark is below 35
    if any(mark < 35 for mark in marks):

        # If at least one mark is below 35, student fails
        status = "Fail"

    else:

        # Otherwise, student passes
        status = "Pass"


    # -----------------------------------------------------
    # CALCULATE GRADE
    # -----------------------------------------------------

    # If the student fails in any subject
    if status == "Fail":

        # Grade is F
        grade = "F"

    # Average 90 or above
    elif average >= 90:

        # Grade is A+
        grade = "A+"

    # Average 80 or above
    elif average >= 80:

        # Grade is A
        grade = "A"

    # Average 70 or above
    elif average >= 70:

        # Grade is B
        grade = "B"

    # Average 60 or above
    elif average >= 60:

        # Grade is C
        grade = "C"

    # Average 50 or above
    elif average >= 50:

        # Grade is D
        grade = "D"

    # Average below 50
    else:

        # Grade is E
        grade = "E"


    # Get the student's name
    # strip() removes unnecessary spaces from beginning and end
    name = name_entry.get().strip()


    # Display the student's result
    result_text.set(
        f"Student: {name}\n"
        f"Total: {total:.2f}\n"
        f"Average: {average:.2f}\n"
        f"Result: {status}\n"
        f"Grade: {grade}"
    )


# ---------------------------------------------------------
# FUNCTION 3: CLEAR / RESET
# ---------------------------------------------------------

def clear_all():

    # Clear the student name Entry box
    name_entry.delete(0, tk.END)

    # Clear the number of subjects Entry box
    subject_entry.delete(0, tk.END)

    # Clear the result displayed on the screen
    result_text.set("")

    # Remove all subject mark Entry widgets
    for widget in marks_frame.winfo_children():
        widget.destroy()

    # Clear the list containing mark Entry widgets
    mark_entries.clear()


# ---------------------------------------------------------
# GUI DESIGN
# ---------------------------------------------------------

# Create the application heading
tk.Label(
    root,
    text="Student Performance Analyzer",
    font=("Arial", 16, "bold")
).pack(pady=10)


# Create Student Name label
tk.Label(
    root,
    text="Student Name"
).pack()


# Create Student Name input box
name_entry = tk.Entry(root)
name_entry.pack()


# Create Number of Subjects label
tk.Label(
    root,
    text="Number of Subjects"
).pack()


# Create Number of Subjects input box
subject_entry = tk.Entry(root)
subject_entry.pack()


# Create Enter Marks button
# When clicked, enter_marks() function will execute
tk.Button(
    root,
    text="Enter Marks",
    command=enter_marks
).pack(pady=8)


# Create a frame to hold subject mark input boxes
marks_frame = tk.Frame(root)
marks_frame.pack()


# Create Calculate Result button
# When clicked, calculate_result() function will execute
tk.Button(
    root,
    text="Calculate Result",
    command=calculate_result
).pack(pady=8)


# StringVar is used to store and update the result text
result_text = tk.StringVar()


# Create a label to display the result
tk.Label(
    root,
    textvariable=result_text,
    justify="left",
    font=("Arial", 11)
).pack(pady=10)


# Create Clear / Reset button
# When clicked, clear_all() function will execute
tk.Button(
    root,
    text="Clear / Reset",
    command=clear_all
).pack(pady=8)


# Start the Tkinter event loop
# This keeps the application window running
root.mainloop()


```
