```python
#QUESTION
#Create a GUI attendance application where students can be added, marked Present/Absent and summarized
#in a report. The GUI should contain Student Name, Add Student, a Listbox, Present and Absent buttons,
#Generate Report, Total Students, Present, Absent and Attendance Percentage.

#ANSWER

import tkinter as tk
from tkinter import messagebox


# ---------------------------------------------------------
# CREATE MAIN WINDOW
# ---------------------------------------------------------

root = tk.Tk()

# Set the title of the application
root.title("Student Attendance Manager")

# Set the window size
root.geometry("600x550")


# ---------------------------------------------------------
# DATA STORAGE
# ---------------------------------------------------------

# Dictionary to store student names and attendance status
#
# Example:
# {
#     "Arun": "Present",
#     "Kumar": "Absent"
# }
students = {}


# ---------------------------------------------------------
# ADD STUDENT
# ---------------------------------------------------------

def add_student():

    # Get the student name from the Entry box
    name = name_entry.get().strip()

    # Check whether the name is empty
    if name == "":
        messagebox.showerror(
            "Error",
            "Enter a student name."
        )
        return

    # Check whether the student already exists
    if name in students:
        messagebox.showwarning(
            "Warning",
            "Student already exists."
        )
        return

    # Add the student with default status
    students[name] = "Not Marked"

    # Add student to the Listbox
    student_list.insert(
        tk.END,
        f"{name} - Not Marked"
    )

    # Clear the Entry box
    name_entry.delete(0, tk.END)


# ---------------------------------------------------------
# MARK PRESENT
# ---------------------------------------------------------

def mark_present():

    # Get the selected student
    selected = student_list.curselection()

    # Check whether a student is selected
    if not selected:
        messagebox.showwarning(
            "Warning",
            "Select a student first."
        )
        return

    # Get the selected student's index
    index = selected[0]

    # Get the student name from the dictionary
    name = list(students.keys())[index]

    # Change attendance status to Present
    students[name] = "Present"

    # Update the Listbox
    refresh_list()


# ---------------------------------------------------------
# MARK ABSENT
# ---------------------------------------------------------

def mark_absent():

    # Get the selected student
    selected = student_list.curselection()

    # Check whether a student is selected
    if not selected:
        messagebox.showwarning(
            "Warning",
            "Select a student first."
        )
        return

    # Get the selected student's index
    index = selected[0]

    # Get the student name
    name = list(students.keys())[index]

    # Change attendance status to Absent
    students[name] = "Absent"

    # Update the Listbox
    refresh_list()


# ---------------------------------------------------------
# REFRESH LISTBOX
# ---------------------------------------------------------

def refresh_list():

    # Remove all existing items
    student_list.delete(0, tk.END)

    # Add updated student information
    for name, status in students.items():

        student_list.insert(
            tk.END,
            f"{name} - {status}"
        )


# ---------------------------------------------------------
# GENERATE REPORT
# ---------------------------------------------------------

def generate_report():

    # Calculate total number of students
    total_students = len(students)

    # Count students marked Present
    present = 0

    # Count students marked Absent
    absent = 0

    # Loop through attendance statuses
    for status in students.values():

        if status == "Present":
            present += 1

        elif status == "Absent":
            absent += 1


    # Calculate attendance percentage
    if total_students > 0:
        attendance_percentage = (
            present / total_students
        ) * 100

    else:
        attendance_percentage = 0


    # Display the attendance report
    report_text.set(
        f"Total Students: {total_students}\n"
        f"Present: {present}\n"
        f"Absent: {absent}\n"
        f"Attendance Percentage: "
        f"{attendance_percentage:.2f}%"
    )


# ---------------------------------------------------------
# CLEAR / RESET
# ---------------------------------------------------------

def clear_all():

    # Remove all students from dictionary
    students.clear()

    # Clear the Listbox
    student_list.delete(0, tk.END)

    # Clear the report
    report_text.set("")

    # Clear the student name Entry
    name_entry.delete(0, tk.END)


# =========================================================
# GUI DESIGN
# =========================================================

# ---------------------------------------------------------
# TITLE
# ---------------------------------------------------------

tk.Label(
    root,
    text="Student Attendance Manager",
    font=("Arial", 16, "bold")
).pack(pady=10)


# ---------------------------------------------------------
# STUDENT NAME
# ---------------------------------------------------------

# Create Student Name label
tk.Label(
    root,
    text="Student Name"
).pack()


# Create Student Name input box
name_entry = tk.Entry(root)

name_entry.pack()


# ---------------------------------------------------------
# ADD STUDENT BUTTON
# ---------------------------------------------------------

tk.Button(
    root,
    text="Add Student",
    command=add_student
).pack(pady=8)


# ---------------------------------------------------------
# STUDENT LISTBOX
# ---------------------------------------------------------

# Create Listbox to display students
student_list = tk.Listbox(
    root,
    width=45,
    height=10
)

student_list.pack(pady=10)


# ---------------------------------------------------------
# PRESENT BUTTON
# ---------------------------------------------------------

tk.Button(
    root,
    text="Present",
    command=mark_present
).pack(pady=3)


# ---------------------------------------------------------
# ABSENT BUTTON
# ---------------------------------------------------------

tk.Button(
    root,
    text="Absent",
    command=mark_absent
).pack(pady=3)


# ---------------------------------------------------------
# GENERATE REPORT BUTTON
# ---------------------------------------------------------

tk.Button(
    root,
    text="Generate Report",
    command=generate_report
).pack(pady=8)


# ---------------------------------------------------------
# REPORT DISPLAY
# ---------------------------------------------------------

# StringVar is used to store the report
report_text = tk.StringVar()


# Label displays the report
tk.Label(
    root,
    textvariable=report_text,
    justify="left",
    font=("Arial", 11)
).pack(pady=10)


# ---------------------------------------------------------
# CLEAR / RESET BUTTON
# ---------------------------------------------------------

tk.Button(
    root,
    text="Clear / Reset",
    command=clear_all
).pack(pady=5)


# ---------------------------------------------------------
# START APPLICATION
# ---------------------------------------------------------

# Start the Tkinter event loop
root.mainloop()
```
